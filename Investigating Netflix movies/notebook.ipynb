## Research Questions

**What was the most frequent movie duration in the 1990s?**

We'll find the most frequently occurring duration for movies released between 1990-1999 and save this as an integer called `duration`.

**How many short action movies were released in the 1990s?**

A movie is considered short if its duration is less than 90 minutes. We'll count the number of short action movies from the 1990s and save this count as `short_movies_count`.

Feel free to experiment further after submitting the project!

```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

```python
# Start coding here! Use as many cells as you like
# the most frequent movie duration in the 1990s
movies_1990s = netflix_df[np.logical_and(netflix_df['release_year'] >=1990 , netflix_df['release_year'] <=1999, netflix_df['type'] == 'Movie')]
durations = np.array(movies_1990s['duration'])
max_count=0
duration=0
for i in durations : 
    count = np.sum(durations==i)
    if count > max_count :
        max_count=count
        duration=i
print(duration)
# A movie is considered short if it is less than 90 minutes.
short_movies = movies_1990s[np.logical_and(movies_1990s['duration'] < 90 , movies_1990s['genre'] == 'Action')]
short_movies_count = short_movies.shape[0]
print(short_movies_count)
```
